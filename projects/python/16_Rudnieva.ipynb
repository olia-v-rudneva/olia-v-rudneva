## Частина 5. Прості цикли

### Завдання 23

Є список `sales_q1`.

За допомогою циклу `for` порахуй суму всіх продажів у `sales_q1`.

### Завдання 37

Додай до DataFrame `df` стовпець `total`.

Він має дорівнювати сумі `sales_q1` і `sales_q2`.

````python
# Завдання 37

df['total'] = df['sales_q1'] + df['sales_q2']
df
````

### Завдання 38

Додай до DataFrame `df` стовпець `growth_pct`.

Формула:

```python
(sales_q2 - sales_q1) / sales_q1 * 100
```

Результат округли до одного знаку після коми.

````python
# Завдання 38

df['growth_pct'] = ((df['sales_q2'] - df['sales_q1']) / df['sales_q1'] * 100).round(1)
df
````

### Завдання 39

Додай до DataFrame `df` стовпець `delta`.

Він має дорівнювати різниці:

```python
sales_q2 - sales_q1
```

````python
# Завдання 39

df['delta'] = df['sales_q2'] - df['sales_q1']
df
````

### Завдання 40

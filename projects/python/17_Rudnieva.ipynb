## Частина 2. Створення нових ознак та аналіз родинних зв'язків



5. Необхідно створити єдиний стовпчик 'relatives', що вказує кількість родичів для кожного пасажира на борту, замість:
- кількість братів/сестер або чоловіків/дружин на борту;
- кількість батьків або дітей на борту;

Булева ознака: стовпчик 'alone'=True, якщо пасажир подорожував один (без родичів на борту).    
Перевір, щоб для одиноких пасажирів новий стовпчик мав значення 0.   
Після створення нового стовпчика, дропни "sibsp", "parch", "alone".     
Порахуй кількість пропущених значень (NaN) у стовпці relatives, збережи результат у змінну та виведи її.

```python
# Завдання 5

df['relatives'] = df['sibsp'] + df['parch']
print(df.loc[df['alone'] == True, 'relatives'].value_counts())
df = df.drop(columns=['sibsp', 'parch', 'alone'])
nan_relatives = df['relatives'].isna().sum()
print("Кількість NaN у стовпчику relatives:", nan_relatives)
display(df)
```

6. Перевір, як часто зустрічається певна кількість родичів в новому стовпці.     
Як результат, створи та відобрази DataFrame з частотою появи кожного значення `relatives`, використовуючи групування та агрегацію.

7. Використовуючи цикл заміни кількість родичів, що перевищує число 5(п'ять) на значення "above 5"(понад п'ять).       
Запиши значення в новий стовпчик ʼrelatives_categoryʼ.     
Як результат, створи та відобрази DataFrame з частотою появи кожного значення `relatives_category`, використовуючи групування та агрегацію.

```python
# Завдання 7
relatives_category = []
for iii in df.relatives:
  if iii <= 5:
    relatives_category.append(str(iii))
  else:
    relatives_category.append('above 5')
df['relatives_category'] = relatives_category
relatives_category = df.groupby('relatives_category').size().reset_index(name='frequency')
display(relatives_category)
```

## Частина 3. Групування та статистика за портами посадки

8. Необхідно вивести на екран статистику по відсотковому показнику пасажирів з кількістю родичів більше 5 відносно загального числа пасажирів для кожного з міст посадки. Для цього:
- порахуй загальну кількість пасажирів в кожному з міст посадки,
- знайди число пасажирів з кількістю родичів більше 5 в кожному з міст посадки (якщо таких пасажирів немає, використай значення 0),
- поділи ці два стовпчики між собою, перетворивши результат у відсотки (ціле число).

Як результат, створи та відобрази DataFrame для кожного міста посадки, що містить усі розраховані показники.

```python
# Завдання 8

embark_town_count = df.groupby('embark_town').size().reset_index(name='amount_passangers')
# display(embark_town_count)
df_above5 = df[df['relatives_category'] == 'above 5']
above5_count = df_above5.groupby('embark_town').size().reset_index(name='amount_passangers_above5')
# display(above5_count)
amount_above5_embark = embark_town_count.merge(above5_count, on='embark_town', how='left')
amount_above5_embark['amount_passangers_above5'] = amount_above5_embark['amount_passangers_above5'].fillna(0)
amount_above5_embark['percentage'] = (amount_above5_embark['amount_passangers_above5'] / amount_above5_embark['amount_passangers'] * 100).astype(int)
display(amount_above5_embark)
```

## Частина 4. Обробка та категоризація віку


9. Заповни відсутні значення віку медіаною.

```python
# Завдання 9

median_age = df['age'].median()
display(median_age)
df['age'].isna().sum()
print("Кількість NaN до заповнення:", df['age'].isna().sum())
df['age'] = df['age'].fillna(median_age)
df['age'].isna().sum()
print("Кількість NaN після заповнення:", df['age'].isna().sum())
display(df)
```

10. Створи новий стовпець, де вік представлено категорією, замість числа (категорії: до 14 років, 14-34 роки, 35-59 років, 60 і більше років). Категорії: `child`, `young`, `adult`, `senior`.   
Як результат, створи та відобрази DataFrame з кількістю пасажирів для кожного значення `age_group`.

```python
from pandas.core.indexes import category
# Завдання 10

df.loc[df['age'] < 14, 'age_group'] = 'child'
df.loc[(df['age'] >= 14) & (df['age'] < 35), 'age_group'] = 'young'
df.loc[(df['age'] >= 35) & (df['age'] < 60), 'age_group'] = 'adult'
df.loc[df['age'] >= 60 , 'age_group'] = 'senior'
df['age_group'] = df['age_group'].astype('category')
# age_group_count = df.groupby('age_group',observed=False).size().reset_index(name='age_group_count').sort_values(by='age_group_count', ascending=False)
age_group_count = df['age_group'].value_counts().reset_index(name='age_group_count')
display(age_group_count)
```

## Частина 5. Аналіз виживання пасажирів

11. Перевір, в якій віковій категорії була найвища смертність. Для цього створи окремий булевий стовпець для пасажирів, які не вижили, використовуючи значення стовпця `alive`.       
Для кожної вікової категорії підрахуй:
- загальну кількість пасажирів;
- кількість пасажирів, які не вижили;
- відсоток пасажирів, які не вижили.    
Як результат, створи та відобрази DataFrame, де для кожної вікової категорії наведені всі розраховані показники.

```python
# Завдання 11

df['not_survived'] = df['alive'] == 'no'
survival_by_age = df.groupby('age_group', observed=False)['not_survived'].agg(total_passengers='count', not_survived_count='sum', not_survived_percentage='mean').reset_index()
survival_by_age['not_survived_percentage'] = (survival_by_age['not_survived_percentage'] *100).round(2)
display(survival_by_age)
```
